```python
from pathlib import Path
import pandas as pd
from krippendorff import alpha as k_alpha
```

```python
base_path = Path("/content/drive/MyDrive/ep_populism_classification/" if COLAB else "../../")
data_path = base_path / "data/annotations"
```

## Prepare data

```python
df = pd.read_excel(data_path / "ep_populism_gold_sample_combined.xlsx", sheet_name="Combined")

variables = {
    "anti-elitism":    ["anti_elitism_lorenz", "anti_elitism_nadia", "anti_elitism_amelie"],
    "people-centrism": ["people_centrism_lorenz", "people_centrism_nadia", "people_centrism_amelie"],
}
```

```python
k_alphas = {}
for var, cols in variables.items():
    mat = df[cols].replace(0.5, float("nan")).T.values
    k_alphas[var] = k_alpha(mat, level_of_measurement="ordinal")  # ← assign here

print(pd.DataFrame.from_dict(k_alphas, orient="index", columns=["k_alpha"]).round(3).to_markdown())
```
