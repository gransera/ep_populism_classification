## Set up file structure

```python
from google.colab import drive
drive.mount('/content/drive')

import os

# Set your root path — change the folder name to whatever you want
REPO_PATH = '/content/drive/MyDrive/ep_populism_classification'

# Create all folders
folders = [
    'notebooks',
    'src',
    'data',
    'models',
    'checkpoints',
    'outputs',
    'cache',
]

for folder in folders:
    os.makedirs(f"{REPO_PATH}/{folder}", exist_ok=True)
    print(f"✓ created {folder}/")
```
